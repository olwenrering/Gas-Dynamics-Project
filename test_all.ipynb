```python
end_list=[]
Omega_list=[]
R_tot_list=[]
for i in range(loop):
    #print(i)
    end_list.append([])
    Omega_list.append([])
    R_tot_list.append([])
    for j in range(len(mdl[i])):
        if i == 0:
            end = int(len(mdl[i][j].glob)/2)
        else:
            end = len(mdl[i][j].glob)-1
        end_list[i].append(end)
        Omega = mdl[i][j].params.w_rot*1e3
        # Omega_list[i].append(Omega)
        R_tot = mdl[i][j].var[end][0][0]/1e2
        R_tot_list[i].append(R_tot)
        
        Omega_array = mdl[i][j].var[end][15] # rad/s
        r = mdl[i][j].var[end][0]
        Omega_list[i].append(r[0]*Omega_array[0]/1e2)
        print(f"total radius of planet = {R_tot/(R_jup.value)} Rjup")
        print(f"total mass of planet = {mdl[i][j].params.mtot:.4f} Msun")
        print(f"initial angular velocity of planet = {Omega/1e3} km/s")
        print(f"actual angular velocity of planet = {Omega_list[i][j]/1e3:.4f} km/s")
        print("\n")
```

```python
def lmix_R(model, i):
    model.calc_vconv('osc', i=i)
    R_tot = model.var[i][0][0]
    Ro = model.vconv/(np.minimum(model.lmix, R_tot)*model.var[i][15])
    return np.minimum(2*model.lmix*Ro**(3/5)*np.pi, R_tot)
```

```python
def compute_B_half(model, i):
    c = 0.63
    T = model.var[i][2]
    rho = model.var[i][4]
    model.calc_vconv('osc', i=i)
    mask = np.zeros(len(T), dtype='int')
    mask[index_top:index_plots] = int(1)
    mask = mask.astype(bool)
    rho_avg = np.mean(rho[mask])
    qc_0 = model.Fconv[mask][0]
    dr = model.var[i][0]#[mask]
    dT_dr = np.gradient(T, dr)
    H_T_array = -T/dT_dr
    mask_2 = np.where(model.Fconv[mask] > 0)
    R_tot = model.var[i][0][0]
    #print(R_tot)
    #r = ((model.var[i][0])/(R_tot))
    index = np.where(np.abs(T - T[-1]/2.78) < 100)
    #print(len(index[0]))
    if len(index[0]) == 0:
        index = np.where(np.abs(T - T[-1]/2.78) < 300)
    #H_T_obs = dr[index][0]
    H_T_obs = H_T_array[mask][mask_2]
    #print(H_T_obs)
    V = (4/3)*np.pi*(R_tot)**3
    dF = (model.Fconv[mask][mask_2]*np.minimum(model.lmix[mask][mask_2], R_tot)/(qc_0*H_T_obs))**(2/3)*(rho[mask][mask_2]/rho_avg)**(1/3)*4*np.pi*dr[mask][mask_2]**2
    mask_3 = np.where(dF > 0)
    F = np.trapezoid(dF[mask_3][::-1], dr[mask][mask_2][mask_3][::-1]) / V 
    print(F)
    #F= np.trapz(dF[mask_3][::-1], dr[mask_2][mask_3][::-1]) / V
    #print(f"F = {F}")
    B = np.sqrt(2*c*F*rho_avg**(1/3)*qc_0**(2/3))
    return B
```

```python
def Alfven(model, i):
    model.calc_vconv('osc', i=i)
    B = compute_B_half(model, i)
    rho = model.var[i][4]
    return B**2/(4*np.pi*rho*model.vconv**2)

```
